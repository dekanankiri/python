### Hanling imbalance dataset
1. Up Sammpling 
2. Down Sampling

```python
import numpy as np
import pandas as pd

# Set the random seed for reproducibilicty
np.random.seed(123)

# Create a dataframe with two classes
n_sample = 1000
class_0_ratio = 0.9
n_class_0 = int(n_sample * class_0_ratio)
n_class_1 = n_sample - n_class_0
```

```python
## CREATE MY DATAFRAME WITH IMBALANCE DATASET
class_0 = pd.DataFrame({
  'feature_1': np.random.normal(loc=0, scale=1,size=n_class_0),
  'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
  'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
  'feature_1':np.random.normal(loc=2, scale=1, size=n_class_1),
  'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
  'target': [1] * n_class_1
})

```

```python
## Creating DataFrame
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
## Upsampling
df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample
df_minority_upsample = resample(df_minority,replace=True, ## sample with replacement
                                n_samples=len(df_majority),
                                random_state=42
)
```

```python
df_minority_upsample.head()
```

```python
df_upsampled = pd.concat([df_majority,df_minority_upsample]).reset_index(drop=True)
```

```python
df_upsampled['target'].value_counts()
```

```python
# Set the random seed for reproducibilicty
np.random.seed(123)

# Create a dataframe with two classes
n_sample = 1000
class_0_ratio = 0.9
n_class_0 = int(n_sample * class_0_ratio)
n_class_1 = n_sample - n_class_0

## CREATE MY DATAFRAME WITH IMBALANCE DATASET
class_0 = pd.DataFrame({
  'feature_1': np.random.normal(loc=0, scale=1,size=n_class_0),
  'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
  'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
  'feature_1':np.random.normal(loc=2, scale=1, size=n_class_1),
  'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
  'target': [1] * n_class_1
})

## Creating DataFrame
df = pd.concat([class_0,class_1]).reset_index(drop=True)

## Check the class distribution
print(df['target'].value_counts())
```

```python
## downsampling
df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample
df_majority_downsample = resample(df_majority,replace=False,
                                n_samples=len(df_minority),
                                random_state=42
)
```

```python
df_majority_downsample.shape
```

```python
df_downsampled = pd.concat([df_minority,df_majority_downsample]).reset_index(drop=True)
df_downsampled
```
