### EDA With Red Wine Data
Data Set Information:

The two dataset are related to red and white variants of the portuguese "Vinho Verde" wine. Due to privacy and logistic issues, only physicochemical (inputs) and sensory (the outputs) variables are available (e.g. there is no data about grape types, wine brand, wine selling price, etc.).

These Datasets can be viewed as classification or regression tasks. The classes are ordered and not balanced (e.g. there are many more normal wines than excellent or poor ones). Outlier detection algorithm could be used to detect the few excelent or poor wines. Also, we are not sure if all input variabes are relevant. So it could be interesting to test feature selection methods.

Attribute Information:

Input variables (based on physicochemical tests):
1. fixed acidity
2. volatile acidity
3. Critic Acid
4. Residual sugar
5. chlorides
6. free sulfur dioxide
7. total sulfur dioxide
8. density
9. ph
10. sulphates
11. alcohol output variable (based on sensory data):
12. Quality (score between 0 and 10)
